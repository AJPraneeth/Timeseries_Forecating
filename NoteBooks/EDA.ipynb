```python
import pandas as pd
```

```python
df=pd.read_csv('Data_download\datasets\yahoo_stock.csv')
```

```python
df.head()
```

```python
df=df.set_index('Date')
```

```python
df.info()
```

```python
import seaborn as sns

import matplotlib.pyplot as plt

# Display basic statistics
print(df.describe())
```

```python
fig, axes = plt.subplots(nrows=2, ncols=6, figsize=(24, 14))

for idx, i in enumerate(df.columns):
    # Plotting the price over time
    ax = axes[0, idx]
    ax.plot(df.index, df[i], label=f'{i} Price')
    ax.set_title(f'{i} Price Over Time')
    ax.set_xlabel('Date')
    ax.set_ylabel(f'{i} Price')
    ax.legend()

    # Plotting the distribution of prices
    ax = axes[1, idx]
    sns.histplot(df[i], bins=50, kde=True, ax=ax)
    ax.set_title(f'Distribution of {i} Prices')
    ax.set_xlabel(f'{i} Price')
    ax.set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
df.corr()
```

```python

# Plotting the correlation matrix
plt.figure(figsize=(10, 6))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```
