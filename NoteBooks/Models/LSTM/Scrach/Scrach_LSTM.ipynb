```python
import torch 
import torch.nn as nn
```

```python
input_size =5
hiden_size = 24
```

```python
#input getaet weights
W_ix = nn.Parameter(torch.Tensor(hiden_size, input_size))
W_ih = nn.Parameter(torch.Tensor(hiden_size, hiden_size))
b_i = nn.Parameter(torch.Tensor(hiden_size))
```

```python
W_ix.shape, W_ih.shape, b_i.shape
```

```python
x_t = torch.randn(input_size)
h_t = torch.randn(hiden_size)
```

```python
i_t= torch.sigmoid(torch.matmul(W_ix, x_t) + torch.matmul(W_ih, h_t) + b_i)
```

```python
i_t_2 = torch.sigmoid(x_t@ W_ix.T+ h_t @ W_ih.T+ b_i)
```

```python
i_t.shape, i_t_2.shape
```

```python
#Forget gate weights
W_fx = nn.Parameter(torch.Tensor(hiden_size, input_size))
W_fh = nn.Parameter(torch.Tensor(hiden_size, hiden_size))
b_f = nn.Parameter(torch.Tensor(hiden_size))
```

```python
f_t = torch.sigmoid(torch.matmul(W_fx, x_t) + torch.matmul(W_fh, h_t) + b_f)
f_t_2 = torch.sigmoid(x_t@ W_fx.T+ h_t @ W_fh.T+ b_f)
```

```python
#cell gate weights
W_gx = nn.Parameter(torch.Tensor(hiden_size, input_size))
W_gh = nn.Parameter(torch.Tensor(hiden_size, hiden_size))
b_g = nn.Parameter(torch.Tensor(hiden_size))
```

```python
g_t= torch.tanh(torch.matmul(W_gx, x_t) + torch.matmul(W_gh, h_t) + b_g)
```

```python
#output gate weights
W_ox = nn.Parameter(torch.Tensor(hiden_size, input_size))
W_oh = nn.Parameter(torch.Tensor(hiden_size, hiden_size))
b_o = nn.Parameter(torch.Tensor(hiden_size))
```

```python
o_t = torch.sigmoid(torch.matmul(W_ox, x_t) + torch.matmul(W_oh, h_t) + b_o)
```

```python
i_t.shape, f_t.shape, g_t.shape, o_t.shape
```

```python
input_gate=i_t* g_t
```

```python
c_prev= torch.randn(hiden_size)
```

```python
forget_gate = f_t * c_prev
```

```python
c_t = forget_gate + input_gate
```

```python
h_t_new= o_t * torch.tanh(c_t)
```
